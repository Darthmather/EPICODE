```python
import pandas as pd

diabetes = pd.read_csv('C:/Users/matte/Downloads/archive/beginner_datasets/diabetes.csv')

dimensione = diabetes.shape
print("Dimensione del dataset:", dimensione)

anteprima = diabetes.head()
print("Anteprima delle prime cinque righe:\n", anteprima)

metadati = diabetes.info()
print("\nMetadati delle colonne:\n", metadati)

descrittori_statistici = diabetes.describe()
print("\nDescrittori statistici del dataset:\n", descrittori_statistici)

fascia_20_29 = diabetes[(diabetes['Age (years)'] >= 20) & (diabetes['age'] < 30)]
fascia_30_39 = diabetes[(diabetes['Age (years)'] >= 30) & (diabetes['age'] < 40)]
fascia_40_50 = diabetes[(diabetes['Age (years)'] >= 40) & (diabetes['age'] < 51)]

media_pressione_20_29 = fascia_20_29['diastolic_pressure'].mean() if not fascia_20_29.empty else None
media_pressione_30_39 = fascia_30_39['diastolic_pressure'].mean() if not fascia_30_39.empty else None
media_pressione_40_50 = fascia_40_50['diastolic_pressure'].mean() if not fascia_40_50.empty else None

print("\nMedia della pressione sanguigna diastolica:")
print(f"Fascia 20-29: {media_pressione_20_29}")
print(f"Fascia 30-39: {media_pressione_30_39}")
print(f"Fascia 40-50: {media_pressione_40_50}")
```

```python
import pandas as pd

insurance = pd.read_csv('C:/Users/matte/Downloads/archive/beginner_datasets/insurance.csv')

dimensione = insurance.shape
print("Dimensione del dataset:", dimensione)

anteprima = insurance.head()
print("Anteprima delle prime cinque righe:\n", anteprima)

nomi_colonne = insurance.columns.tolist()
print("Nomi delle colonne:", nomi_colonne)

medie_charges_region = insurance.groupby('region')['charges'].mean()
print("\nMedie di charges rispetto a region:\n", medie_charges_region)

varianze_region = insurance.groupby('region')['charges'].var()
print("\nVarianze di charges rispetto a region:\n", varianze_region)

medie_charges_smoker = insurance.groupby('smoker')['charges'].mean()
print("\nMedie di charges rispetto a smoker:\n", medie_charges_smoker)

medie_charges_sex = insurance.groupby('sex')['charges'].mean()
print("\nMedie di charges rispetto a sex:\n", medie_charges_sex)

descrittori_bmi = insurance['bmi'].describe()
print("\nDescrittori statistici di BMI:\n", descrittori_bmi)

quartili_bmi = insurance['bmi'].quantile([0.25, 0.5, 0.75])

charges_quartili = []
for q in quartili_bmi:
    charges_media = insurance[insurance['bmi'] <= q]['charges'].mean()
    charges_quartili.append(charges_media)

print("\nMedia di charges rispetto ai diversi quartili di BMI:", charges_quartili)
```

```python
import pandas as pd
import random

file_path = 'C:/Users/matte/Downloads/Dataset_Lezioni/dataset_es4/Mappa-dei-pub-circoli-locali-in-Italia.csv'
pub_data = pd.read_csv(file_path, encoding='latin1', sep=';')

numero_dati = pub_data.shape[0]
print("Numero totale di dati:", numero_dati)

metadati = pub_data.info()
print("\nMetadati:\n", metadati)

primo_elemento = pub_data.iloc[0]
print("\nPrimo elemento:\n", primo_elemento)

ultimo_elemento = pub_data.iloc[-1]
print("\nUltimo elemento:\n", ultimo_elemento)

elemento_casuale = pub_data.sample(1)
print("\nElemento a caso:\n", elemento_casuale)

anni_inserimento = pub_data['Anno inserimento'].unique()
print("\nAnni di inserimento presenti:", anni_inserimento)

pub_data['Tipo'] = pub_data['Nome'].str.extract(r'(?i)\b(Enoteca|Cantina|Vigna|Bottega|Wine|Chianti|Bacchus|Vinoteca)\b', expand=False)

wine_houses = pub_data[pub_data['Tipo'].notna()]
numero_wine_houses = wine_houses.shape[0]
nomi_wine_houses = wine_houses['Nome'].unique()

print("\nNumero di wine houses:", numero_wine_houses)
print("Nomi delle wine houses:", nomi_wine_houses)

tipi_count = pub_data['Tipo'].value_counts()
print("\nDistribuzione dei tipi di locali:")
print(tipi_count)
```
