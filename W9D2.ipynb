```python
def fattori_primi(numero):
    fattori = []
    divisore = 2
    
    while numero > 1:
        while numero % divisore == 0:
            fattori.append(divisore)
            numero //= divisore
        divisore += 1

    return fattori

numero_input = int(input("Inserisci un numero: "))

fattori = fattori_primi(numero_input)

print(f"I fattori di {numero_input} sono: {fattori}")
```

```python
N = int(input("Inserisci il numero di potenze da calcolare (N): "))
K = int(input("Inserisci la base (K): "))

potenze = []
exp = 0

while exp < N:
    potenza = K**exp
    potenze.append(potenza)
    exp += 1

print("Lista potenze:", potenze)
```
