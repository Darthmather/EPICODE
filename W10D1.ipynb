```python
import pandas as pd

vendite = [2000, 5500, 7200, 4320, 1280, 1900, 2500, 3900, 6410, 8150, 7100, 5350]
vendite = pd.Series(vendite)

print(f"Vendita mensile massima: {vendite.max()}")
print(f"Vendita mensile minima: {vendite.min()}")

sells_5k = vendite[vendite > 5000]

print(f"Vendite maggiori di 5k: {sells_5k.values}")
print(f"5k counter: {sells_5k.count()}")

sells_3k = vendite[vendite < 3000]
print(f"Vendite minori di 3k: {sells_3k.values}")

sells_mean = vendite.mean()
print(f"Media delle vendite: {sells_mean.round()}")
```

```python
import pandas as pd

stipendi = pd.Series([100, 200, 300, 400, 500, 600, 700, 800, 900, 1000])

doubled_stipendi = pd.Series(index=stipendi.index)
for i in range(len(stipendi)):
    doubled_stipendi[i] = stipendi[i] * 2

print(f"Stipendi raddoppiati con ciclo for: \n{doubled_stipendi}")

masking_stipendi = stipendi * 2
print(f"Stipendi raddoppiati con masking: \n{masking_stipendi}")
```

```python
def big3(lista):
    # Caso lista vuota
    if not lista:
        return []
    
    # Caso lista con meno di 3 elementi
    if len(lista) < 3:
        return sorted(lista, reverse=True)
    
    # Caso almeno 3 elemnti
    big3 = sorted(lista, reverse=True)[:3]
    
    return big3
```

```python
def mean_number(lista, K):
    #Filtraggio <> per K
    num_filter = [num for num in lista if num >= K]
    
    #Check sul criterio
    if not num_filter:
        print(f"Nessun numero maggiore o uguale a {K} trovato nella lista.")
        return None
    
    # Calcola la media dei numeri filtrati
    return sum(num_filter) / len(num_filter)
```

```python
def censura(nums):
    # Ciclo per ogni numero nella lista
    for num in nums:
        # Stampa una riga di asterischi pari al valore del numero
        print('*'*num)

set_nums = [5, 2, 3, 4]
censura(set_nums)
```

```python
def cf_check(cf_list):
    cf_95 = []
    
    for cf in cf_list:
        #Check sull'anno e pure sulla lunghezza del CF
        if len(cf) == 16 and "95" in cf:
            cf_95.append(cf)
            
            cognome = cf[:3]
            nome = cf[3:6]
            
            print(f"Codice Fiscale Valido: {cf}")
            print(f"Caratteri Cognome: {cognome}")
            print(f"Caratteri Nome: {nome}")
            print("-" * 30)
    
    return cf_95

cf_list = [
        "ABCDEF95G01A123B", 
        "GHIJKL91M02A321C", 
        "MNOPQR89S03A456D", 
        "STUVWX95Z04A654E", 
        "XYZABC01D05A789F", 
        "DEFGHI95J06A987G"
]
cf_check(cf_list)
```

```python
def first_gen(studenti, corsi, edizioni):
    first_studenti = []
    
    for i, edizione in enumerate(edizioni):
        if edizione == 1:
            studente = studenti[i]
            corso = corsi[i]
            
            first_studenti.append(studente)
            
            print(f"Studente: {studente}")
            print(f"Corso: {corso}")
            print("-" * 20)
    
    return first_studenti

studenti = ["Alex", "Bob", "Cindy", "Dan", "Emma", "Faith", "Grace", "Henry"]
corsi = ["Cybersecurity", "Data Analyst", "Backend", "Frontend", "Data Analyst", "Backend", "Frontend", "Cybersecurity"]
edizioni = [1, 2, 3, 2, 2, 1, 3, 3]

first_gen(studenti, corsi, edizioni)
```

```python
def convertion(prices):
    new_prices = []
    
    for price in prices:
        try:
            if '€' in price:
                new_price = price.replace('€', '$')
                new_prices.append(new_price)
        except Exception as e:
            print(f"Errore durante la conversione: {e}")
            new_prices.append(price)
    
    return new_prices

prices = ["100 €", "200 €", "500 €","10 €", "50 €", "70 €"]
convertion(prices)
```

```python
def div_team(students):
    squadra_pari = [student for i, student in enumerate(students) if i % 2 == 0]
    squadra_dispari = [student for i, student in enumerate(students) if i % 2 == 1]
    
    return squadra_pari, squadra_dispari

students = ["Alex", "Bob", "Cindy", "Dan", "Emma", "Faith", "Grace", "Henry", "Isabelle", "John"]
div_team(students)
```

```python
def analyze_money(revenues):
    months = [
    "Gennaio", "Febbraio", "Marzo", "Aprile",
    "Maggio", "Giugno", "Luglio", "Agosto",
    "Settembre", "Ottobre", "Novembre", "Dicembre"
    ]
    
    sum_builder = 0.0  # accumulatore 

    for i, revenue in enumerate(revenues):
        month = months[i]

        if i == 0:                              # primo mese: non esiste media precedente
            print(f"{month:>10}: {revenue:>4} €  (primo mese)")
        else:
            mean = sum_builder/i         # media dei valori precedenti
            diff = abs(revenue-mean)
            state = "MAGGIORE" if revenue > mean else "MINORE"

            # Formattazione allineata per leggibilità
            print(
                f"{month:>10}: {revenue:>4} €  "
                f"│ media precedenti: {mean:6.2f} €  "
                f"│ {state:7} di {diff:5.2f} €"
            )

        sum_builder += revenue            # aggiorna l’accumulatore


# ----------------------------------------------------------------------
revenues = [100, 90, 70, 40, 50, 80, 90, 120, 80, 20, 50, 50]
analyze_money(revenues)
```
